```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import drive

filepath = "/content/D3.csv"
 # If you want to test this be sure to link this properly
df = pd.read_csv(filepath)

index = np.arange(0, 100)
x1 = df.values[:, 0]
x2 = df.values[:, 1]
x3 = df.values[:, 2]
y = df.values[:, 3]
m = len(y)
n = len(x1)

#print(x1)
#print(x2)
#print(x3)
#print(y)

cf = 10
lr = 0.02
p1 = 0
p1i = 0
p2 = 0
p2i = 0
p3 = 0
p3i = 0


def cost_function(res, y):
  return np.sum((res - y) ** 2) / (2 * m)
def plot_result_to_target(index, result, y, x):
  plt.figure(figsize=(10, 6))
  plt.scatter(index, y, label='Index vs Y', alpha=0.7)
  plt.scatter(index, result, label='Index vs Result', alpha=0.7)
  plt.scatter(index, x, label='Index vs X', alpha=0.7)
  plt.show()
def gradient(result, y, x):
  temp = [0, 0]
  temp[0] = np.sum((result - y) * x) / m
  temp[1] = np.sum(result - y) / m
  return temp
def linregression(data, param, inter, y):
  grad = [1, 1]
  iteration = 0
  while abs(grad[0]) > 0.01 or abs(grad[1]) > 0.01:
    result = data * param + inter
    cf = cost_function(result, y)
    #Only plot if you want to see the progress at each step, can be computationally intensive
    grad = gradient(result, y, data)
    param = param - lr * grad[0]
    inter = inter - lr * grad[1]
    if iteration % 100 == 0:
      print(f"Iteration: {iteration}")
      print(f"Current parameter: {param:.6f}")
      print(f"Current intercept: {inter:.6f}")
      print(f"Current cost: {cf:.6f}")
    iteration += 1
  plot_result_to_target(index, result, y, data)
  print(f"Total iterations: {iteration}")
  iteration = 0


linregression(x1, p1, p1i, y)
linregression(x2, p2, p2i, y)
linregression(x3, p3, p3i, y)
```
